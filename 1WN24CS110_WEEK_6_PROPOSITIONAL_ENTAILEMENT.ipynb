```python

from itertools import product

def evaluate(expr, model):
    # Replace logical symbols with their truth values
    for symbol in sorted(model, key=len, reverse=True):
        expr = expr.replace(symbol, str(model[symbol]))

    expr = expr.replace("AND", "and")
    expr = expr.replace("OR", "or")
    expr = expr.replace("NOT", "not")
    expr = expr.replace("IMPLIES", "<=")

    return eval(expr)


def tt_entails(KB, alpha, symbols):
    print("\nTRUTH TABLE")
    print("-" * 65)
    print(" | ".join(symbols) +
          " | KB | alpha | KB => alpha")
    print("-" * 65)

    entails = True

    # Generate all possible truth assignments
    for values in product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))

        kb_value = evaluate(KB, model)
        alpha_value = evaluate(alpha, model)

        # KB entails alpha if no row has KB=True and alpha=False
        implication = (not kb_value) or alpha_value

        row = ["T" if model[s] else "F" for s in symbols]
        row += [
            "T" if kb_value else "F",
            "T" if alpha_value else "F",
            "T" if implication else "F"
        ]

        print(" | ".join(row))

        if kb_value and not alpha_value:
            entails = False

    print("-" * 65)

    if entails:
        print("Result: KB entails alpha (TRUE)")
    else:
        print("Result: KB does not entail alpha (FALSE)")


# Get input from the user
symbols = input(
    "Enter propositional symbols (space-separated): "
).split()

KB = input(
    "Enter Knowledge Base (use and, or, not): "
)

alpha = input(
    "Enter query (alpha): "
)

tt_entails(KB, alpha, symbols)
```
